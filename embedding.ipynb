```python
import tensorflow
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sent = ['the glass of milk',
        'the glass of juice',
        'the cup of tea',
        'I am a good boy',
        'I am a good developer',
        'understand the meaning of words',
        'your videos are good']
```

```python
sent
```

```python
# Define the vocabulary size
voc_size = 10000
```

```python
# One Hot Representation for every word
one_hot_representation = [one_hot(words, voc_size)for words in sent]
one_hot_representation
```

```python
# Word Embedding Representation
from tensorflow.keras.layers import Embedding
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
import numpy as np
```

```python
# The number of input should be same but our dataset has variable word length
sent_length = 8
embedded_docs = pad_sequences(one_hot_representation, padding = 'pre', maxlen= sent_length)
print(embedded_docs)
```

```python
# feature representation
dim= 10
```

```python
model = Sequential()
model.add(Embedding(voc_size, dim, input_length= sent_length))
model.compile('adam', 'mse')
```
